```python
import streamlit as st
from pulp import LpMaximize, LpProblem, LpVariable
from pulp import LpStatus
import numpy
```

```python
### **Python Code Using PuLP**


# Define the problem
problem = LpProblem("Maximize_Profit_and_Leads", LpMaximize)

# Define decision variables
x = LpVariable("Direct_Advertising", lowBound=0)  # Amount spent on direct ads
y = LpVariable("Indirect_Advertising", lowBound=0)  # Amount spent on indirect ads

# Objective function: Maximize profit
problem += 5 * x + 3 * y, "Total_Profit"

# Constraints
problem += x + y <= 100000, "Budget_Constraint"              # Total budget
problem += x >= 0.4 * (x + y), "Direct_Minimum"              # Direct ads at least 40%
problem += y <= 50000, "Indirect_Cap"                       # Indirect ads max $50,000
problem += (8000 * x / 1000) + (5000 * y / 1000) >= 700000, "Viewer_Constraint"  # Minimum viewers

# Solve the problem
problem.solve()

# Print results
# print(f"Status: {problem.status}, {LpProblem().status[problem.status]}")
# print(f"Optimal profit: ${problem.objective.value():,.2f}")
# print(f"Amount spent on direct advertising: ${x.value():,.2f}")
status_descriptions = LpProblem().status  # Assuming it's a list or dictionary

print(f"Status: {problem.status}, {status_descriptions[problem.status]}") 
print(f"Optimal profit: ${problem.objective.value():,.2f}") 
print(f"Amount spent on direct advertising: ${x.value():,.2f}")
print(f"Amount spent on indirect advertising: ${y.value():,.2f}")
print(f"Estimated viewers generated: {8000 * x.value() / 1000 + 5000 * y.value() / 1000:,.0f}")
```

Explanation of the Code
Diminishing Returns:
Different formulas for calculating subscribers based on spending thresholds.
A piecewise approach was modeled using factors a

b.Constraints:
Fixed setup costs were added to the budget constraint.
Indirect advertising had a minimum spend constraint.
Direct advertising spend was capped at 60% of the total budget.
Optimization Goal:
Maximize subscribers while adhering to all constraints.

```python
from pulp import LpMaximize, LpProblem, LpVariable

# Constants for diminishing returns
a = 0.02  # Diminishing returns factor for direct ads
b = 0.015  # Diminishing returns factor for indirect ads

# Define the problem
problem = LpProblem("OTT_Advertising_Optimization", LpMaximize)

# Decision variables
x = LpVariable("Direct_Advertising", lowBound=0)  # Amount spent on direct ads
y = LpVariable("Indirect_Advertising", lowBound=0)  # Amount spent on indirect ads

# Objective function: Subscribers from direct and indirect ads
direct_subscribers = (
    50 * x / 1000 if x <= 100000 else 50 * 100 + a * (x - 100000) ** 0.5
)
indirect_subscribers = (
    30 * y / 1000 if y <= 75000 else 30 * 75 + b * (y - 75000) ** 0.6
)

problem += direct_subscribers + indirect_subscribers, "Total_Subscribers"

# Constraints
problem += x + y + 10000 + 7500 <= 200000, "Budget_Constraint"  # Budget limit
problem += y >= 50000, "Indirect_Minimum"  # Minimum spend on indirect ads
problem += x <= 0.6 * (x + y), "Direct_Channel_Limit"  # Direct ads limit

# Solve the problem
problem.solve()

# Extract results
direct_spend = x.value()
indirect_spend = y.value()
total_subscribers = problem.objective.value()

# Display results
print(f"Optimal spending on direct advertising: ${direct_spend:,.2f}")
print(f"Optimal spending on indirect advertising: ${indirect_spend:,.2f}")
print(f"Total subscribers gained: {total_subscribers:,.0f}")
```
